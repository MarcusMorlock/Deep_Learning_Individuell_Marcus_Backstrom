Individuell inlämningsuppgift – Deep Learning

Uppgift: Klassificera ansiktsuttryck med neurala nätverk
I den här uppgiften ska du använda neurala nätverk för att bygga en modell som kan klassificera ansiktsuttryck från bilder.
Målet är att du ska förstå hur en deep learning-modell:
byggs
tränas
utvärderas
används för att göra prediktioner på ny data
Det finns inte en “rätt” lösning – du förväntas testa, utvärdera och justera din modell utifrån hur den presterar.
Uppgiften är praktisk och fokuserar på att du själv ska testa och förstå hur modellen fungerar.


Data
Du ska använda datasetet nedan som innehåller gråskalebilder av ansikten med storlek 48x48 pixlar.

Viktigt
Datasetet är relativt stort.
Börja alltid med en mindre del av datan för att säkerställa att din kod och modell fungerar.
Du behöver inte använda hela datasetet för att bli godkänd.
Att datasetet är stort innebär att träning kan ta tid. I praktiken behöver man ofta balansera mellan modellens kvalitet och hur lång tid det tar att träna och testa olika lösningar. 
Det är därför vanligt att börja med mindre datamängder när man testar modeller, och sedan skala upp när allt fungerar.


Uppdrag
Du ska bygga en modell som kan:
lära sig från bilddata
skilja mellan olika ansiktsuttryck
göra prediktioner på nya bilder


Vad du ska göra
Arbeta iterativt - du kommer behöva gå tillbaka och justera tidigare steg.

1. Förstå datan
Målet är att du ska få en överblick över datan innan du börjar modellera.
Undersök till exempel:
vilka klasser som finns
hur många bilder som finns
om vissa klasser verkar vanligare än andra

2. Förbered datan
Målet är att göra datan användbar för en neural nätverksmodell.
Datasetet är redan uppdelat i tränings- och testdata där varje klass har en egen mapp.

Du behöver:
läsa in bilderna
förstå hur datan är uppdelad och hur den ska användas i träning och testning
skala/normalisera bilderna
Fundera på:
behöver du använda alla klasser?
behöver du begränsa datamängden i början?

3. Bygg en modell
Målet är att skapa en modell som kan lära sig mönster i bilddata.
Du behöver:
använda convolutional layers (CNN)
skapa en modell som är tillräckligt flexibel för att lära sig skillnader mellan klasser
Du behöver själv ta ställning till:
hur många lager modellen ska ha
hur stora lagren ska vara
vilka aktiveringsfunktioner som är lämpliga
Testa dig fram och justera modellen utifrån hur den presterar.
Om modellen är för enkel kan den få svårt att lära sig tillräckligt bra.

4. Träna modellen
Målet är att träna modellen på ett sätt som ger rimliga resultat.
Du behöver:
träna modellen på träningsdata
använda validation under träningen
observera hur modellen utvecklas över tid
Fundera på:
blir modellen bättre ju längre du tränar?
finns det tecken på overfitting?

5. Utvärdera modellen
Målet är att förstå hur bra modellen fungerar på ny data.
Du behöver:
utvärdera modellen på testdata
använda minst en metric (t.ex. accuracy)
tolka resultatet

6. Gör prediktioner på ny data
Målet är att använda modellen som ett “program”.
Du ska skapa kod där du kan:
ladda in en ny bild
förbereda bilden på samma sätt som träningsdatan
använda modellen för att göra en prediktion
visa vilken klass modellen tror att bilden tillhör
Bilden du använder för prediktion ska inte vara en bild från träningsdatan. Du kan till exempel använda en egen bild eller en bild från testdatan

7. Analysera resultatet
Besvara följande i din notebook:
Hur bra fungerar modellen?
Ser du tecken på overfitting eller underfitting?
Hur skiljer sig modellens resultat mellan träningsdata och testdata?
Vad verkar påverka resultatet mest?
Vilka begränsningar har din modell?

8. Reflektion
Besvara kort:
Vad var svårast i uppgiften?
Vad lärde du dig?
Vad hade du gjort annorlunda om du började om?
Vilket betyg tycker du att din inlämning motsvarar, G eller VG?
Motivera kort genom att hänvisa till vilka krav i uppgiften du tycker att du uppfyllt.
Självreflektionen påverkar inte betyget i sig, men hjälper dig att visa att du förstår uppgiftens krav och ditt eget arbete.

Krav för Godkänd (G)
För att bli godkänd ska du:
bygga en fungerande modell baserad på CNN
träna och utvärdera modellen på ett korrekt sätt
visa att du kan göra prediktioner på ny data
genomföra en grundläggande analys av resultatet
visa förståelse för centrala begrepp (t.ex. overfitting)

Krav för Väl Godkänd (VG)
För VG ska du visa en djupare förståelse genom att:

Jämföra modeller
testa flera modellvarianter
undersöka hur olika val påverkar resultatet

Förbättra modellen
testa förändringar som påverkar prestanda
 (t.ex. fler lager, dropout, fler epochs)

Göra en djupare analys
jämföra resultat mellan modeller
tolka träningskurvor
förklara varför modellerna beter sig olika

Föra en kritisk diskussion
diskutera modellens begränsningar
resonera kring datamängd och generalisering
reflektera kring när deep learning är ett bra val


Frivillig bonus (ej krav för betyg)
Om du vill kan du:
bygga en enkel applikation (t.ex. med Streamlit eller Gradio)
testa modellen på egna bilder
använda modellen i realtid via webbkamera
